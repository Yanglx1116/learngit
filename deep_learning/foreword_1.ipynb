数据：训练数据、测试数据
训练数据集用于拟合模型参数，测试数据集用于评估拟合的模型。 
当一个模型在训练集上表现良好，但不能推广到测试集时，这个模型被称为过拟合（overfitting）的。 
模型：定义假设空间的函数集合
损失函数：衡量模型预测与真实值的差异
优化算法：寻找最优模型参数
